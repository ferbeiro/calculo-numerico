QUESTÃO 5

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Sistema linear A x = b
A = np.array([
    [10, -1,  2],
    [-1, 11, -1],
    [ 2, -1, 10]
], dtype=float)

b = np.array([6, 22, -10], dtype=float)

# Aproximacao inicial x^(0) = (0, 0, 0)
x = np.array([0.0, 0.0, 0.0])

# Criterio de parada e limite de seguranca
tol = 1e-5
max_iter = 1000

# Armazenar resultados de cada iteracao
resultados = []

# Metodo de Gauss-Seidel
for k in range(1, max_iter + 1):
    x_anterior = x.copy()

    # x^(k): usa y e z da iteracao anterior
    x[0] = (6 + x_anterior[1] - 2*x_anterior[2]) / 10

    # y^(k): usa o x atualizado e o z anterior
    x[1] = (22 + x[0] + x_anterior[2]) / 11

    # z^(k): usa os x e y atualizados
    x[2] = (-10 - 2*x[0] + x[1]) / 10

    # Erro: norma infinito entre iteracoes consecutivas
    erro = np.max(np.abs(x - x_anterior))

    resultados.append([k, x[0], x[1], x[2], erro])

    if erro < tol:
        break

# Questao 5: numero de iteracoes e solucao aproximada
print("QUESTAO 5 — RESULTADO")
print("Numero de iteracoes:", k)
print(f"Solucao aproximada: x = {x[0]:.10f}, y = {x[1]:.10f}, z = {x[2]:.10f}")
print(f"Erro final = {erro:.10e}")
print("Criterio atendido:", erro < tol)
```

QUESTÃO 6

```python
# Questao 6: tabela completa das iteracoes
tabela = pd.DataFrame(
    resultados,
    columns=["k", "x(k)", "y(k)", "z(k)", "Erro infinito"]
)

display(tabela.round(10))
```

```python
# Questao 7(a): grafico de x(k), y(k) e z(k)
# Esta celula e independente: pode ser executada sozinha.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

A = np.array([
    [10, -1,  2],
    [-1, 11, -1],
    [ 2, -1, 10]
], dtype=float)
b = np.array([6, 22, -10], dtype=float)

x = np.array([0.0, 0.0, 0.0])
tol = 1e-5
max_iter = 1000
resultados_7a = []

for k in range(1, max_iter + 1):
    x_anterior = x.copy()
    x[0] = (6 + x_anterior[1] - 2*x_anterior[2]) / 10
    x[1] = (22 + x[0] + x_anterior[2]) / 11
    x[2] = (-10 - 2*x[0] + x[1]) / 10
    erro = np.max(np.abs(x - x_anterior))
    resultados_7a.append([k, x[0], x[1], x[2], erro])
    if erro < tol:
        break

tabela_7a = pd.DataFrame(
    resultados_7a,
    columns=["k", "x(k)", "y(k)", "z(k)", "Erro infinito"]
)

plt.figure(figsize=(8, 5))
plt.plot(tabela_7a["k"], tabela_7a["x(k)"], marker="o", label="x(k)")
plt.plot(tabela_7a["k"], tabela_7a["y(k)"], marker="s", label="y(k)")
plt.plot(tabela_7a["k"], tabela_7a["z(k)"], marker="^", label="z(k)")
plt.xlabel("Numero da iteracao k")
plt.ylabel("Valor da aproximacao")
plt.title("Questao 7(a) - Variaveis por iteracao")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
