## 2. Método da eliminação de Gauss

Sistema:

$$
\begin{cases}
2x+y-z=8\\
-3x-y+2z=-11\\
-2x+y+2z=-3
\end{cases}
$$

Matriz aumentada inicial:

$$
\left[
\begin{array}{ccc|c}
2&1&-1&8\\
-3&-1&2&-11\\
-2&1&2&-3
\end{array}
\right]
$$

### Etapa 1

Aplicando

$$L_2\leftarrow 2L_2+3L_1$$

e

$$L_3\leftarrow L_3+L_1,$$

obtemos

$$
\left[
\begin{array}{ccc|c}
2&1&-1&8\\
0&1&1&2\\
0&2&1&5
\end{array}
\right].
$$

### Etapa 2

Aplicando

$$L_3\leftarrow L_3-2L_2,$$

temos

$$
\left[
\begin{array}{ccc|c}
2&1&-1&8\\
0&1&1&2\\
0&0&-1&1
\end{array}
\right].
$$

### Substituição regressiva

Da terceira linha:

$$-z=1 \Rightarrow z=-1.$$

Da segunda linha:

$$y+z=2,$$

logo,

$$y-1=2\Rightarrow y=3.$$

Da primeira linha:

$$2x+y-z=8,$$

então

$$2x+3-(-1)=8,$$

$$2x+4=8,$$

$$x=2.$$

### Resultado

$$\boxed{x=2,\quad y=3,\quad z=-1}$$

## 3. Número de soluções


- **Solução única:** há pivô em todas as variáveis.
- **Infinitas soluções:** existe pelo menos uma variável livre e não aparece uma contradição.
- **Nenhuma solução:** aparece uma linha do tipo $[0\ 0\ \cdots\ 0\mid c]$, com $c\neq0$.

### 3(a)

$$
\begin{cases}
x+y=3\\
2x+2y=6
\end{cases}
$$

Matriz aumentada:

$$
\left[
\begin{array}{cc|c}
1&1&3\\
2&2&6
\end{array}
\right]
$$

Aplicando $L_2\leftarrow L_2-2L_1$:

$$
\left[
\begin{array}{cc|c}
1&1&3\\
0&0&0
\end{array}
\right].
$$

A segunda equação é redundante. Temos uma variável livre, portanto existem **infinitas soluções**.

Podemos escrever:

$$x=3-y.$$

Assim,

$$\boxed{\text{infinitas soluções}}$$

### 3(b)

$$
\begin{cases}
x+y=3\\
2x+2y=7
\end{cases}
$$

Matriz aumentada:

$$
\left[
\begin{array}{cc|c}
1&1&3\\
2&2&7
\end{array}
\right].
$$

Aplicando $L_2\leftarrow L_2-2L_1$:

$$
\left[
\begin{array}{cc|c}
1&1&3\\
0&0&1
\end{array}
\right].
$$

A última linha representa

$$0=1,$$

que é uma contradição.

Logo,

$$\boxed{\text{nenhuma solução}}$$

### 3(c)

$$
\begin{cases}
x+y=3\\
2x-y=0
\end{cases}
$$

Matriz aumentada:

$$
\left[
\begin{array}{cc|c}
1&1&3\\
2&-1&0
\end{array}
\right].
$$

Aplicando $L_2\leftarrow L_2-2L_1$:

$$
\left[
\begin{array}{cc|c}
1&1&3\\
0&-3&-6
\end{array}
\right].
$$

Da segunda linha:

$$-3y=-6\Rightarrow y=2.$$

Substituindo na primeira:

$$x+2=3\Rightarrow x=1.$$

Portanto,

$$\boxed{\text{solução única: }(x,y)=(1,2)}$$

## 4. Implementação numérica — Eliminação de Gauss



```python
import numpy as np

def eliminacao_gauss(A, b):
    # Converte os dados para ponto flutuante
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)

    # Forma a matriz aumentada [A | b]
    M = np.column_stack((A, b))

    n = len(b)

    # Eliminação de Gauss
    for k in range(n - 1):
        if abs(M[k, k]) < 1e-12:
            raise ValueError("Pivô nulo. O método sem pivoteamento não pode continuar.")

        for i in range(k + 1, n):
            fator = M[i, k] / M[k, k]
            M[i, k:] = M[i, k:] - fator * M[k, k:]

    # Substituição regressiva
    x = np.zeros(n)

    for i in range(n - 1, -1, -1):
        soma = np.dot(M[i, i + 1:n], x[i + 1:n])
        x[i] = (M[i, n] - soma) / M[i, i]

    return x, M


# Sistema do exercício 2
A = [
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
]

b = [8, -11, -3]

solucao_gauss, matriz_final = eliminacao_gauss(A, b)

print("Matriz escalonada:")
print(matriz_final)

print("\nSolução pelo método de Gauss:")
print(solucao_gauss)
```

### Comparação com `numpy.linalg.solve`



```python
solucao_numpy = np.linalg.solve(np.array(A, dtype=float),
                                np.array(b, dtype=float))

print("Solução pelo método de Gauss:   ", solucao_gauss)
print("Solução pelo numpy.linalg.solve:", solucao_numpy)

print("\nAs soluções são iguais?",
      np.allclose(solucao_gauss, solucao_numpy))
```

### Interpretação

Os dois métodos devem produzir a mesma solução:

$$\boxed{(x,y,z)=(2,3,-1)}.$$

Isso confirma a implementação do algoritmo para este sistema. O método implementado reproduz numericamente o procedimento de eliminação de Gauss usado na resolução manual.

## 5. Aplicação — Modelo de balanço térmico

O sistema fornecido é

$$
\begin{bmatrix}
4&-1&0\\
-1&4&-1\\
0&-1&3
\end{bmatrix}
\begin{bmatrix}
T_1\\T_2\\T_3
\end{bmatrix}
=
\begin{bmatrix}
15\\10\\10
\end{bmatrix}.
$$

Isso corresponde a

$$
\begin{cases}
4T_1-T_2=15\\
-T_1+4T_2-T_3=10\\
-T_2+3T_3=10
\end{cases}
$$

### 5(a) Resolução manual

Matriz aumentada:

$$
\left[
\begin{array}{ccc|c}
4&-1&0&15\\
-1&4&-1&10\\
0&-1&3&10
\end{array}
\right].
$$

Para evitar frações logo no início, podemos usar operações equivalentes.

Aplicando

$$L_2\leftarrow4L_2+L_1,$$

obtemos

$$
\left[
\begin{array}{ccc|c}
4&-1&0&15\\
0&15&-4&55\\
0&-1&3&10
\end{array}
\right].
$$

Agora, para eliminar o termo $-1$ da terceira linha, fazemos

$$L_3\leftarrow15L_3+L_2.$$

Então:

$$
\left[
\begin{array}{ccc|c}
4&-1&0&15\\
0&15&-4&55\\
0&0&41&205
\end{array}
\right].
$$

Da terceira linha:

$$41T_3=205$$

$$\boxed{T_3=5}.$$

Da segunda linha:

$$15T_2-4T_3=55.$$

Substituindo $T_3=5$:

$$15T_2-20=55,$$

$$15T_2=75,$$

$$\boxed{T_2=5}.$$

Da primeira linha:

$$4T_1-T_2=15.$$

Substituindo $T_2=5$:

$$4T_1-5=15,$$

$$4T_1=20,$$

$$\boxed{T_1=5}.$$

Portanto,

$$\boxed{T_1=T_2=T_3=5}.$$

### 5(b) Implementação em Python


```python
A_termica = [
    [4, -1, 0],
    [-1, 4, -1],
    [0, -1, 3]
]

b_termica = [15, 10, 10]

temperaturas, matriz_termica = eliminacao_gauss(A_termica, b_termica)

print("Matriz escalonada:")
print(matriz_termica)

print("\nTemperaturas calculadas:")
print(f"T1 = {temperaturas[0]:.2f}")
print(f"T2 = {temperaturas[1]:.2f}")
print(f"T3 = {temperaturas[2]:.2f}")
```

### Interpretação do resultado

A solução numérica é

$$
\boxed{T_1=T_2=T_3=5}.
$$

Portanto, para o sistema fornecido, as três temperaturas desconhecidas apresentam o mesmo valor de **5 unidades de temperatura**.

A solução obtida pelo código coincide com a resolução manual.
